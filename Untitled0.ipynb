```python
from sentence_transformers import SentenceTransformer
import faiss
import numpy as np
from langchain_community.llms import Replicate
import os
from google.colab import userdata
```

```python
api_token = userdata.get('REPLICATE_api_token')
os.environ["REPLICATE_API_TOKEN"] = api_token
model = "ibm-granite/granite-3.2-8b-instruct"
output = Replicate(
model=model,
replicate_api_token=api_token,
)
```

```python
user_query = input("Ask a question: ")
initial_response = output.invoke(user_query)
print("\nInitial Response (No RAG):\n")
print(initial_response)
```

```python
from google.colab import files
uploaded = files.upload()
doc_path = list(uploaded.keys())[0]
with open(doc_path, 'r', encoding='utf-8') as f:
 document_text = f.read()
def split_into_chunks(text, chunk_size=300, overlap=50):
 chunks = []
 start = 0
 while start < len(text):
  end = start + chunk_size
  chunks.append(text[start:end])
  start += chunk_size - overlap
 return chunks
chunks = split_into_chunks(document_text)
print(f"Document split into {len(chunks)} chunks.")
```

```python
embed_model = SentenceTransformer('all-MiniLM-L6-v2')
embeddings = embed_model.encode(chunks)
dimension = embeddings.shape[1]
# Initialize FAISS index
index = faiss.IndexFlatL2(dimension)
index.add(embeddings)
# Keep chunks for later retrieval
stored_chunks = chunks
print(f"Stored {len(stored_chunks)} embeddings in FAISS index.")
```

```python
# Embed the user's query
query_embedding = embed_model.encode([user_query])

# Search the FAISS index for the most similar chunks
k = 3 # Number of nearest neighbors to retrieve
distances, indices = index.search(query_embedding, k)

# Retrieve the actual chunks
retrieved_chunks = [stored_chunks[i] for i in indices[0]]

# Combine the retrieved chunks with the user's query
context = "\n".join(retrieved_chunks)
prompt_with_context = f"""Answer the question based on the provided context. If the answer is not in the context, say 'I cannot answer this question based on the provided information.'

Context:
{context}

Question: {user_query}
Answer:"""

# Generate the RAG response using the LLM
rag_response = output.invoke(prompt_with_context)

print("\nRag Response:\n")
print(rag_response)
```
